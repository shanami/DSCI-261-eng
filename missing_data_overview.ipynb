```python
# Check if there is a null value (no value)
wdi.isna()
```

```python
# Check if there is no null value (has a value)
wdi.notna()
```

```python
# Get the dataframe with/without null values, then sum
wdi.isna().sum()
```

```python
# Count the number of occurrences of each value in the alcohol_consumption_per_capita column
wdi['alcohol_consumption_per_capita'].value_counts()
```

```python
# NaN: not a number
# Counts the number of times each value appears in the alcohol_consumption_per_capita column, including missing cells (NaN)

# wdi['alcohol_consumption_per_capita'] selects the average per capita alcohol consumption column (a Series)
# value_counts() counts how many times each different value appears, sorted from most to least
# dropna=False is the most important point. By default, value_counts() ignores NaN
# When dropna=False is set, NaN is counted as a normal value and appears as a separate row in the results
wdi['alcohol_consumption_per_capita'].value_counts(dropna=False)
```

```python
wdi.isna().sum(axis=1)
```

```python
# Count the number of empty rows, then sum
# Axis = 1 takes the row (axis = 0 takes the column)
num_missing_by_row = wdi.isna().sum(axis=1)
```

```python
# Show lines with empty values
num_missing_by_row > 0
```

```python
# number of empty lines
(num_missing_by_row > 0).sum()
```
