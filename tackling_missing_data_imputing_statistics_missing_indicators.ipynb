# fillna()

```python
import pandas as pd
wdi = pd.read_pickle('wdi.pkl')

# Get the names of the numeric columns and the non-numeric columns
numeric_cols = wdi.select_dtypes(include='number').columns
cat_cols = wdi.select_dtypes(exclude='number').columns
```

```python
# Returns a Series: each column contains one value, with the index being the column name
# Mean and median both ignore NaN when calculating
wdi[numeric_cols].mean()
```

```python
wdi[numeric_cols].median()
```

```python
# fillna takes a Series as the fill value: pandas match by column name, so the NaN of each column is replaced by the average of that column
# This command only returns a new table to view, wdi remains unchanged
wdi[numeric_cols].fillna(wdi[numeric_cols].mean())
```

```python
# Similar to the above but assigned to wdi, so the change is real
# This is how pure pandas are equivalent to SimpleImputer(strategy='mean')
wdi[numeric_cols] = wdi[numeric_cols].fillna(wdi[numeric_cols].mean())
```

```python
# Count the values, including NaN
# After the cell above, this column no longer has NaN, and the cells that were missing hours all have the same value (which is the average), so that value will appear unusually many times
# This is a familiar consequence of filling in the average: narrowing the distribution
wdi['alcohol_consumption_per_capita'].value_counts(dropna=False)
```

```python
# For a text column, `describe()` returns 4 rows: count (number of cells that are not missing), unique (number of different values), top (most frequent value, i.e., mode), freq (number of times top appears)
wdi[cat_cols].describe()
```

```python
# .loc['top'] selects the correct top row, outputs Series: index is the column name, value is the most frequent value
wdi[cat_cols].describe().loc['top']
```

```python
# Save it to the variable most_freq
most_freq = wdi[cat_cols].describe().loc['top']
```

```python
# Similar to step 4 but for text columns: each column is filled with its own mode, then written back to wdi
# The last line only displays the result
# This is equivalent to SimpleImputer(strategy='most_frequent')
wdi[cat_cols] = wdi[cat_cols].fillna(most_freq)
wdi[cat_cols]
```

```python
wdi['country_category'].value_counts(dropna=False)
```

```python
wdi.info()
```

# SimpleImputer

```python
wdi = pd.read_pickle('wdi.pkl')
numeric_cols = wdi.select_dtypes(include='number').columns
cat_cols = wdi.select_dtypes(exclude='number').columns
```

```python
from sklearn.impute import SimpleImputer
simple_imp_mean = SimpleImputer(strategy='mean')
```

```python
imputed_num = simple_imp_mean.fit_transform(wdi[numeric_cols])
```

```python
print(imputed_num.shape)
print(wdi[numeric_cols].shape)
```

```python
wdi[numeric_cols] = imputed_num
```

```python
wdi['compulsory_education_years'].value_counts(dropna=False)
```

```python
wdi.info()
```

```python
# simple_imp_med = SimpleImputer(strategy='median')
```

```python
# Fill missing cells in the text column with the most frequently occurring value (mode) of that column, using sklearn's SimpleImputer
# This is sklearn's version of the describe().loc['top'] + fillna(most_freq) step

# strategy='most_frequent': each column is filled with the most frequent value of that column
# Unlike strategy='constant' (a general constant like 'MISSING'), here each column has its own fill value, derived from the data at the time of fitting
# No need for fill_value, as the fill value is automatically found by the imputer
# The _freq name indicates that this imputer uses the most frequent value
simple_imp_freq = SimpleImputer(strategy='most_frequent')

# fit: For each column in wdi[cat_cols], find the most frequently occurring value (ignore NaN)
# These values ​​are stored in simple_imp_freq.statistics_
# transform: Replace all NaN with the mode of the corresponding column
# The result is a 2D numpy array of object type stored in imputed_cat
# At this point, wdi is not changed; to write it back, you must assign: wdi[cat_cols] = imputed_cat
imputed_cat = simple_imp_freq.fit_transform(wdi[cat_cols])
```

```python
wdi[cat_cols] = imputed_cat
```

```python
wdi.info()
```

```python
wdi['country_category'].value_counts(dropna=False)
```

# Missing Indicator

```python
wdi = pd.read_pickle('wdi.pkl')
```

```python
# Create an imputer that fills in NaN cells with the average of the column, and adds flag columns (0/1) to indicate which cells were originally missing

# strategy='mean': Each NaN cell is replaced with the average of that column (calculated from the data at the time of fitting, ignoring NaN). Only usable for numeric columns
# add_indicator=True: In addition to the filled columns, the imputer appends indicator columns to the right of the result. The value is 1 if the original cell is missing, 0 if there is data
simple_imp_ind = SimpleImputer(strategy='mean', add_indicator=True)
```

```python
wdi['compulsory_education_years']
```

```python
type(wdi['compulsory_education_years'])
```

```python
wdi[['compulsory_education_years']]
```

```python
type(wdi[['compulsory_education_years']])
```

```python
# Apply the imputer with the flag column to only one column, compulsory_education_years: fill in NaN with the average, and create an additional column indicating which cells are missing

# wdi[['compulsory_education_years']] has two pairs of [[ ]] so the result is a 1-column DataFrame. This is important: sklearn's fit_transform requires 2D data
# If you write wdi['compulsory_education_years'](Series, 1D), sklearn will report an error: Expected 2D array, got 1D array instead
# simple_imp_ind is the imputer created in the previous step: SimpleImputer(strategy='mean', add_indicator=True)
# fit_transform learns the average of the column (ignore NaN), fills in the missing cells, and then appends the flag column
imputed_ind = simple_imp_ind.fit_transform(wdi[['compulsory_education_years']])
```

```python
# imputed_ind is the numpy array with 2 columns from the previous step
# [:20] retrieves the first 20 rows
# Each row has the format [filled value, flag]
imputed_ind[:20]
```

```python
imputed_ind.shape
```

```python
# The left side is a 2-column list, the right side is a 2-column array, so pandas assigns the first column of the array to the first column and the second column to the second column

# Compulsory_education_years already exists, so it's overwritten with the data already filled in
# Compulsory_education_years_missing doesn't exist, so pandas creates a new column with the name you set. This is the flag column 0/1

# This command actually fixes wdi. It only runs when the array has exactly 2 columns, meaning this column has NaN during fit
# If the column doesn't have NaN, the array only has 1 column and pandas reports a column mismatch error
wdi[['compulsory_education_years', 'compulsory_education_years_missing']] = imputed_ind
```

```python
# Display the first 20 rows of those two columns
# You should see: rows with flags equal to 1 have the same compulsory_education_years value (both are averages), while rows with flags equal to 0 retain the original value
wdi[['compulsory_education_years', 'compulsory_education_years_missing']].head(20)
```
