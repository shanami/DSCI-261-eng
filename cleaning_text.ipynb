```python
taxi = pd.read_csv('nyc_federal_taxi.csv')
```

```python
taxi.head()
```

```python
taxi['PU_Address'].head(10)
```

```python
taxi['PU_Address'] = taxi['PU_Address'].str.lower()
taxi['PU_Address'].head(10)
```

```python
taxi['PU_Address'] = taxi['PU_Address'].str.strip() # cắt bỏ phần đầu/đuôi/giữa bị khoảng trống
taxi['PU_Address'].head(10)
```

```python
taxi['PU_Address_length'] = taxi['PU_Address'].str.len() # xác định độ dài
taxi['PU_Address_length'].head() # PU_Address_length là đặt cái tên cột mới trong DataFrame taxi, là giá trị độ dài của PU_Address (int)
```

```python
# draw graph
taxi['PU_Address_length'].hist(bins=50)
```

```python
# Show dataframe with 2 columns
# View the 10 longest pick-up addresses in the dataset
# .nlargest(n=10, columns='PU_Address_length') Sort by PU_Address_length column in descending order and get the first 10 rows
taxi[['PU_Address', 'PU_Address_length']].nlargest(n=10, columns='PU_Address_length')
```

```python
# Check if the word "airport" is in PU_Address
taxi['PU_Address'].str.contains('airport')
```

```python
# Set the variable taxi_airport to store the entire dataframe containing the PU_Address column with the word "airport"
# taxi['PU_Address'].str.contains('airport') checks each cell in the PU_Address column for the string containing the word "airport". The result is a Series of True/False, with each row of taxi corresponding to a value
# taxi[ ... ] uses that True/False Series as a boolean mask: only keeping rows that contain True
# taxi_airport = ... saves the filtered DataFrame to a new variable. This variable has all the columns of taxi, just fewer rows
taxi_airport = taxi[taxi['PU_Address'].str.contains('airport')]

# Count the number of times each different value appears in the column, and sort from most to least
taxi_airport['PU_Address'].value_counts()
```

```python
# string processing function; get the first character of each address in the PU_Address column (the one above)
taxi['PU_Address'].str[0]
```

```python
# Get the first two characters of each address in the PU_Address column
# .str[0:2] is the syntax for slicing a string
# where the starting position is counted, the ending position is not
# so [0:2] takes the characters at positions 0 and 1
taxi['PU_Address'].str[0:2].head()
```

```python
# Extract characters from positions 0 to 4 (the first five characters)
# If no number is placed before the colon, the default is 0
taxi['PU_Address'].str[:5]
```

```python
# show the last 3 characters from the bottom up
taxi['PU_Address'].str[-3:]
```

```python
# Replace ny with new york
taxi['PU_Address'] = taxi['PU_Address'].str.replace(' ny;', ' new york;')
```
